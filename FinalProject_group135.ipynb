## Abstract

Due to a majority of largely successful basketball players frequently being of considerable stature, it is thought that there is a potential relationship between basketball players' physical traits and basketball performance. In this study, a look at any potential relationship between the relative wingspan of NBA players and their overall performance through performance metrics such as wingspan, robust algorithm using player tracking and on/off ratings (RAPTOR), daily adjusted and regressed kalman optimized (DARKO), luck-adjusted player estimate using a box prior regularized on-off (LEBRON), true shooting percentage, on-off, daily plus-minus (DPM), minutes played, and minutes per game. Upon examining these variables and their relationship to relative wingspan, we found that only the offensive and defensive versions of LEBRON, DPM, and RAPTOR appeared to have a correlation to relative wingspan.

## Research Question

Does the relative wingspan (wingspan minus height) of an NBA player affect their overall performance? This project will focus on NBA seasons from 2018 to 2022. We select this five-season window because advanced metrics such as RAPTOR, LEBRON, DARKO, True Shooting and On-Off are consistently available during this period, and draft combine wingspan and current-season measurements can be matched to active NBA players. ## Background and Prior Work

Prior research has examined the relationship between body proportions and athletic performance. Monson et al. (2018) find that variations in limb proportions are associated with elite athletic success, suggesting that longer wingspans may confer biomechanical advantages. Similarly, Kaazempur-Mofrad (2022) argues that longer wingspans enhance defensive potential in basketball while potentially hindering shooting efficiency. These findings imply that wingspan may have heterogeneous effects across offensive and defensive dimensions of performance.

However, existing literature primarily focuses on biomechanical mechanisms or isolated performance indicators. There has been limited examination of how wingspan relates to modern composite advanced metrics that integrate offensive and defensive contributions into single numerical estimates.

Most of the advanced metrics used in this study, such as RAPTOR, LEBRON, and DARKO, are composite metrics that combine both offensive and defensive components into a single numerical value. These metrics incorporate box score statistics, on/off impact, and regression-based modeling to estimate a player’s overall contribution. Since wingspan is believed to influence defensive actions such as shot contesting, deflections, and rim protection, we expect it to primarily affect the defensive components of these metrics. In contrast, True Shooting Percentage (TS%) measures scoring efficiency and does not directly incorporate defensive performance.

By examining relative wingspan in relation to advanced metrics during the 2018–2022 NBA seasons, our study extends prior research by evaluating whether the defensive advantages suggested in the literature translate into measurable effects within contemporary performance models.

We could not find studies detailing tracking different advanced metrics such as LEBRON, DARKO and RAPTOR and using wingspan to track any correlations in performance.

## Hypothesis


H1a: Relative wingspan is positively associated with defensive advanced metrics (e.g., Defensive RAPTOR, Defensive LEBRON, Defensive DARKO).

H1b: Relative wingspan is negatively associated with scoring efficiency and general offensive metrics (e.g., True Shooting Percentage, Offensive RAPTOR, Offensive LEBRON, Offensive DARKO).

H1c: Relative wingspan has a positive association with composite advanced metrics (e.g., overall RAPTOR, LEBRON, DARKO, On-Off), reflecting the net balance between defensive gains and potential offensive trade-offs.

H0: Relative wingspan has no statistically significant association with these performance metrics during the 2018–2022 NBA seasons.

We believe that there is evidence to suggest that wingspan may negatively affect offensive capabilities, but positively affect offensive capabilities, and that the net tips in favor of a positive correlation in performance overall, as evidenced by our sources.

### Complete Data Source

We have a complete data source where we merged our different data from each of the 5 seasons, which were in turn created by merging all of the relevant statistics for each year by player. It has the players, total number of games played in the season, total minutes, the LEBRON variations, DARKO variations, RAPTOR variations, TS and On-Off, as well as height, wingspan and relative wingspan. This final csv table was created in "Final_Data_Generator.ipynb". First, we loaded all of the cleaned datasets for every metric. We then merged each dataset based on the season, using on="Player", and each season was the result of merging 5 different datasets, specifically the LEBRON, RAPTOR, DARKO, TS and On-Off. We then added a Season column that was just the season, for each final year dataset. We then concatenated the 2018-2022 final datasets into one big dataset. We then merged the wingspan dataset after renaming the "name" column in the final wingspan dataset to "Player", and merged on "Player". Our rationale was that we were going to treat every season's result for one player as one data entry. ### Data Models and Analysis - Analyzing with Math and Models

We will show the results from the Pearson correlation and OLS analysis. The single-variable OLS shows the results from a Pearson correlation, which are the correlation and p-value. This will show whether there is a weak or strong relationship between relative winsgpan and our advanced metrics, and once we find a correlation, we check to see if it is meaningful, or just fluff based on other potential factors.

#### Pearson and OLS

```python
single_relative_model_names = []
single_relative_models = []
for f in features:
    formula = f'Q("{f}") ~ relative'
    model = smf.ols(formula, data=final_result).fit()
    single_relative_models.append(model)
    single_relative_model_names.append(f'{f}_ols')

relative_r_sqr = [] 
relative_p_val = []
relative_coef = []
for model in single_relative_models:
    relative_r_sqr.append(model.rsquared)
    relative_p_val.append(model.pvalues['relative'])
    relative_coef.append(model.params['relative'])
relative_summary = pd.DataFrame({
    'feature': features,
    'correlation': fr_relative_corr[features],
    'coef': relative_coef,
    'r_squared': relative_r_sqr,
    'p_value': relative_p_val
})
relative_summary['p_value'] = relative_summary['p_value'].round(5)
relative_summary.drop(columns=['feature'], inplace=True)
```

```python
print('           OLS summary using Relative as X\n')
print(relative_summary)
```

```python
#OLS Model for Wingspan, Height and Relative
wingspan_and_height_and_relative_model_names = []
wingspan_and_height_and_relative_models = []
for f in features:
    formula = f'Q("{f}") ~ relative + wingspan_inches + height_inches'
    model = smf.ols(formula, data=final_result).fit()   
    wingspan_and_height_and_relative_models.append(model)
    wingspan_and_height_and_relative_model_names.append(f'{f}_ols')
wingspan_and_height_and_relative_r_sqr = [] 
wingspan_and_height_and_relative_p_val = []
wingspan_and_height_and_relative_coef = []
for model in wingspan_and_height_and_relative_models:
    wingspan_and_height_and_relative_r_sqr.append(model.rsquared)
    wingspan_and_height_and_relative_p_val.append(model.pvalues['relative'])
    wingspan_and_height_and_relative_coef.append(model.params['relative'])
wingspan_and_height_and_relative_summary = pd.DataFrame({
    'feature': features,
    'correlation': fr_relative_corr.loc[features],
    'coef': wingspan_and_height_and_relative_coef,
    'r_squared': wingspan_and_height_and_relative_r_sqr,
    'p_value': wingspan_and_height_and_relative_p_val
})
wingspan_and_height_and_relative_summary['p_value'] = wingspan_and_height_and_relative_summary['p_value'].round(5)
wingspan_and_height_and_relative_summary.drop(columns=['feature'], inplace=True)
```

While the models are sometimes statistically significant for raptor_offense, dpm, and LEBRON, their R-squared values are still low, so we can say that the variable effects are not very large for these metrics.

Perhaps in the future, researchers can shift their focus to these subjects to truly understand what leads to succesful basketball performance, which will indubitably lead to better basketball.
